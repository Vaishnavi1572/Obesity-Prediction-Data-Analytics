# **Inital EDA**

```python

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Vaishnavi1572/Obesity-Prediction-Data-Analytics/main/ObesityDataSet_raw_and_data_sinthetic%202.csv"

df = pd.read_csv(url)

df.head()
```

```python
# Dataset information
df.info()
```

```python
# describe
df.describe()
```

```python
# shape
df.shape
```

```python
# null checks
df.isnull().sum()
```

```python
# duplicate check
df.duplicated().sum()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Check duplicate records
df.duplicated().sum()
```

```python
#Remove Duplicates
df = df.drop_duplicates()
# Verify duplicates removed
df.duplicated().sum()
```

```python
#Rename Columns
df.rename(columns={
    'family_history_with_overweight':'Family_History',
    'NObeyesdad':'Obesity_Level'
}, inplace=True)

#Check Updated Columns
df.columns
```

```python
#Treat Outliers
import seaborn as sns
import matplotlib.pyplot as plt

numerical_cols = ['Age','Weight','FAF']

for col in numerical_cols:
    plt.figure(figsize=(5,2))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
#Check Skewness
df[numerical_cols].skew()
```

```python
#Convert Data Types
#Check Data Types
df.dtypes
```

```python
#Convert Categorical Columns
categorical_cols = [
    'Gender',
    'Family_History',
    'FAVC',
    'CAEC',
    'SMOKE',
    'SCC',
    'CALC',
    'MTRANS',
    'Obesity_Level'
]

for col in categorical_cols:
    df[col] = df[col].astype('category')
```

```python
#Verify Conversion
df.dtypes
```

```python
#Feature Transformations
df['BMI'] = df['Weight'] / (df['Height'] ** 2)
```

```python
#Create Age Group Feature
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0,18,30,45,60,100],
    labels=['Teen','Young Adult','Adult','Middle Age','Senior']
)
```

```python
#View New Features
df[['Age','Age_Group','Height','Weight','BMI']].head()
```

```python
#obesity Level Distribution (Countplot)
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.countplot(y=df['Obesity_Level'])
plt.title('Distribution of Obesity Levels')
plt.show()
```

**Interpretation:**

Obesity Type I has the highest number of individuals in the dataset, indicating that early-stage obesity is the most common condition and should be the primary focus for preventive health interventions.

```python
#Age Distribution (Histogram)
plt.figure(figsize=(6,4))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Age Distribution')
plt.show()
```

**Interpretation:**

The age distribution is concentrated among young adults (around 18–30 years),indicating that obesity awareness and prevention efforts should target this population segment.

```python
#Scatter Plot (Age vs Weight)
plt.figure(figsize=(6,4))
sns.scatterplot(x='Age', y='Weight', data=df)
plt.title('Age vs Weight')
plt.show()
```

**Interpretation:**

Weight varies significantly across age groups, indicating that obesity risk is influenced by multiple lifestyle and behavioral factors rather than age alone.

```python
#Bar Plot (Average Weight by Obesity Level)
plt.figure(figsize=(8,4))
sns.barplot(x='Obesity_Level', y='Weight', data=df)
plt.title('Average Weight by Obesity Level')
plt.xticks(rotation=45)
plt.show()
```

**Interpretation:**

Average weight increases consistently across obesity levels, indicating that weight is a strong predictor of obesity risk and should be a key focus for early health interventions.

```python
plt.figure(figsize=(8,6))

numerical_cols = ['Age','Height','Weight','FCVC','NCP','CH2O','FAF','TUE']

sns.heatmap(df[numerical_cols].corr(),
            annot=True,
            cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```

**Interpretation:**

The heatmap shows that weight has the strongest positive relationship with height (0.46), while most other factors have weak correlations, indicating that obesity is influenced by a combination of lifestyle and physical factors rather than a single variable.

```python
#Grouped Analysis
plt.figure(figsize=(8,5))
sns.barplot(x='Age_Group', y='Weight', hue='Gender', data=df)
plt.title('Average Weight by Age Group and Gender')
plt.show()
```

**Interpretation:**

Weight is generally higher among males across age groups, with adult males representing the most important target group for obesity intervention efforts.

```python
sns.pairplot(df[['Age','Height','Weight']])
plt.show()
```

**Interpretation:**



**Interpretation:**

Weight shows a stronger relationship with height than age, indicating that physical characteristics play a greater role in obesity assessment than age alone.

```python
pd.crosstab(
    [df['Age_Group'], df['Gender']],
    df['Obesity_Level']
).plot(
    kind='bar',
    stacked=True,
    figsize=(12,6)
)

plt.title('Obesity Level by Age Group and Gender')
plt.ylabel('Count')
plt.show()
```

**Interpretation:**

Obesity is most prevalent among young adults across both genders, making this age group the primary target for awareness campaigns and weight-management initiatives.
